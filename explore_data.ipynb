## 2. Load Data Files

```python
# Load all parquet files
statcast = pd.read_parquet("data/statcast_2025.parquet")

print("✓ Statcast data loaded!")
print(f"  Shape: {statcast.shape}")
print(f"  Date range: {statcast['game_date'].min()} to {statcast['game_date'].max()}")

# Try to load pitcher profiles if it exists and is not empty
try:
    if os.path.getsize("data/pitcher_profiles.parquet") > 0:
        pitcher_profiles = pd.read_parquet("data/pitcher_profiles.parquet")
        print(f"✓ Pitcher profiles loaded: {pitcher_profiles.shape}")
except:
    print("⚠ Pitcher profiles not yet generated")
```

## 3. Statcast Data Exploration

```python
print("Shape:", statcast.shape)
print("\nColumn Data Types:")
print(statcast.dtypes)
print("\nFirst few rows:")
statcast.head()
```

```python
print("\nStatistical Summary:")
statcast.describe()
```

## 4. Pitch Type Distribution

```python
print("Pitch Type Distribution:")
print(statcast['pitch_type'].value_counts())
print("\nPercentage:")
print(statcast['pitch_type'].value_counts(normalize=True) * 100)
```

## 5. Game State Analysis

```python
print("Game State Counts:")
print(f"Total pitches: {len(statcast)}")
print(f"\nBalls distribution:")
print(statcast['balls'].value_counts().sort_index())
print(f"\nStrikes distribution:")
print(statcast['strikes'].value_counts().sort_index())
print(f"\nOuts distribution:")
print(statcast['outs_when_up'].value_counts().sort_index())
```

```python
print("\nKey Features for Model Training:")
print(statcast[['balls', 'strikes', 'outs_when_up', 'inning', 'release_speed', 'pitch_type']].describe())
```
